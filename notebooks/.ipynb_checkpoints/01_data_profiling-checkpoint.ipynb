## Data Quality & Cleaning: TotalCharges

During data profiling, `TotalCharges` was imported as an object instead of a numeric type due to 11 empty whitespace strings (" ").

All 11 affected records correspond to new customers with a `tenure` of 0 months and `Churn` = No. Because these accounts have not completed their first billing cycle, setting `TotalCharges` to 0.0 accurately reflects their status while maintaining data integrity.

```python
# check blank TotalCharges records

blank_totalCharges = df[
    df["TotalCharges"].astype(str).str.strip() == ""
]
```

```python
blank_totalCharges
```

```python
blank_totalCharges[
    [
        "customerID",
        "tenure",
        "Contract",
        "MonthlyCharges",
        "TotalCharges",
        "Churn"
    ]
]
```

```python
blank_totalCharges["tenure"].value_counts()
```

```python
blank_totalCharges["MonthlyCharges"]
```

```python
blank_totalCharges["Contract"].value_counts()
```

```python
blank_totalCharges["Churn"].value_counts()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)
```

```python
print("Original rows:", len(df))
```

```python
print("Customer IDs:", df["customerID"].nunique())
```

## TotalCharges Analysis

Inspection shows that all 11 records with missing TotalCharges share a `tenure` of 0, `Churn`=No, and active MonthlyCharges. This confirms the missing values belong exclusively to new accounts without a billing history.

## Validation

`TotalCharges` was successfully converted to numeric with missing values imputed as 0.0. Cleaning validation verifies zero remaining nulls, an unchanged record count (7,043), and preserved customer ID uniqueness.

```python
df_clean = df.copy()
```

```python
df_clean.columns.tolist()
```

```python
# checking spaces in text colums

txt_columns = df_clean.select_dtypes(include="object").columns

whitespace_issue = {}

for col in txt_columns:
    count = (
        df_clean[col]
        .astype(str)
        .ne(df_clean[col].astype(str).str.strip())
        .sum()
    )
    if count > 0:
        whitespace_issue[col] = count

if whitespace_issue:
    print("Whitespace issues found")
    print(whitespace_issue)
else:
    print("No whitespace detected")
```

```python
for col in txt_columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()
```

```python
print("Missing TotalCharges:", df_clean["TotalCharges"].isna().sum())
```
